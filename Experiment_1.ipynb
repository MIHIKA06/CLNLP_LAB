#Experiment-1: Data Analysis


##Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/CNNLP_LAB/employee_information_100.csv')
```

```python
avg_salary = df.groupby('Department')['Salary'].mean()
```

```python
plt.figure(figsize=(8, 5))
avg_salary.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Salary by Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Salary', fontsize=12)

plt.grid(axis='y', linestyle='--')

plt.tight_layout()
plt.show()
```

##Experiment 1.2: Display the number of employees working in each department.


```python
dept_counts = df['Department'].value_counts()
print("Number of employees per department:")
print(dept_counts)
```

```python
plt.figure(figsize=(8, 5))
dept_counts.plot(kind='bar', color='teal', edgecolor='black')
plt.title('Employee Count by Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.grid(axis='y', linestyle='--')


plt.tight_layout()
plt.show()
```

##Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['pink', 'red'])
plt.title('Percentage of Male and Female Employees')
plt.ylabel('')
plt.show()
```

##Experiment 1.4: Draw a histogram showing salary distribution.


```python
plt.hist(df['Salary'], bins=5, color='lavender', edgecolor='black')

plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Number of Employees')

plt.show()
```

##Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.


```python
plt.scatter(df['Experience_Years'], df['Salary'], color='purple', alpha=0.7)

plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

##Experiment 1.6: Display the top 10 employees with the highest salary.


```python
top_10 = df.nlargest(10, 'Salary')

print(top_10[['Employee_ID', 'Name', 'Department', 'Salary']])
```

##Experiment 1.7: Find the highest salary in every department.


```python
max_salary = df.groupby('Department')['Salary'].max()

print(max_salary)
```

##Experiment 1.8: Display employees whose salary is greater than the overall average salary.


```python
avg_salary = df['Salary'].mean()

above_avg = df[df['Salary'] > avg_salary]

print(f"Overall Average Salary: {avg_salary:.2f}\n")
print(above_avg[['Employee_ID', 'Name', 'Department', 'Salary']])
```

##Experiment 1.9: Calculate the average years of experience in each department.


```python
avg_exp = df.groupby('Department')['Experience_Years'].mean().reset_index()

print(avg_exp.to_string(index=False))
```

##Experiment 1.10: Display the distribution of employee ages using a histogram.


```python
plt.hist(df['Age'], bins=5, color='pink', edgecolor='black')

plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.show()
```
