##Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
1.Convert text to lowercase. (Also display how many uppercase?)

2.Remove punctuation.        (Also display how many different types of punctuations?)

3.Remove numbers.            (Also display how many numbers?)

4.Remove extra whitespace. (How much extra whitespace is present?)

5.Display the cleaned text.     


```python
import string

with open('/content/drive/MyDrive/CNNLP_LAB/Experiment-2/2.1_text_data.txt', "r") as f:
    text = f.read()

uppercase_count = sum(1 for c in text if c.isupper())
text_lower = text.lower()

found_puncts = set(c for c in text_lower if c in string.punctuation)
text_no_punct = text_lower.translate(str.maketrans("", "", string.punctuation))

digits_count = sum(1 for c in text_no_punct if c.isdigit())
text_no_nums = "".join(c for c in text_no_punct if not c.isdigit())

extra_spaces = len(text_no_nums) - len(" ".join(text_no_nums.split()))
cleaned_text = " ".join(text_no_nums.split())

print("Uppercase letters count:", uppercase_count)
print("Unique punctuation types found:", len(found_puncts), found_puncts)
print("Digits count:", digits_count)
print("Extra whitespace count:", extra_spaces)
print("\nCleaned Text:\n", cleaned_text)
```

##Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


```python
with open( "/content/drive/MyDrive/CNNLP_LAB/Experiment-2/2.2_tokenization_data.txt", "r") as f:
    text = f.read()
```

```python
import re

words = re.findall(r"\w+", text)

sentences = [s.strip() for s in re.split(r"[.!?]+", text) if s.strip()]

print("Words:", words)
print("Sentences:", sentences)
```

```python
import nltk

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

from nltk.tokenize import sent_tokenize, word_tokenize

words = word_tokenize(text)
sentences = sent_tokenize(text)

print("Words:", words)
print("Sentences:", sentences)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

words = [token.text for token in doc]
sentences = [sent.text.strip() for sent in doc.sents]

print("Words:", words)
print("Sentences:", sentences)
```

##Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens.
(Also, display Stop Words Separately).


```python
nltk.download("stopwords", quiet=True)
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt", quiet=True)
nltk.download("stopwords", quiet=True)

with open("/content/drive/MyDrive/CNNLP_LAB/Experiment-2/2.3_clean_data.txt", "r") as f:
    text = f.read()

tokens = word_tokenize(text)
stop_words = set(stopwords.words("english"))

removed_stopwords = [w for w in tokens if w.lower() in stop_words]
filtered_tokens = [w for w in tokens if w.lower() not in stop_words]

print("Original Tokens:\n", tokens)
print("\nDetected Stop Words:\n", set(removed_stopwords))
print("\nFiltered Tokens:\n", filtered_tokens)
```
