##Part 4.2: Term-Frequency Analysis (Without Toolkits)

1)Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

2)Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re

with open(DATASET_PATH, 'r', encoding='utf-8') as f:
    text = f.read()

cleaned_text = re.sub(r'[^\w\s]', '', text.lower())
words = cleaned_text.split()
```

```python
frequency_dict = {}
for word in words:
    if word:
        frequency_dict[word] = frequency_dict.get(word, 0) + 1

with open('output_4_2_tf.csv', 'w', encoding='utf-8') as f:
    f.write("Term,Frequency\n")
    for term, freq in frequency_dict.items():
        f.write(f"{term},{freq}\n")

print("Saved TF analysis to output_4_2_tf.csv")
```

```python
sorted_terms = sorted(frequency_dict.items(), key=lambda x: x[1], reverse=True)

print("\n--- Top 10 Most Frequent Terms (Without Toolkit) ---")
print(f"{'Term':<15} | {'Frequency':<10}")
print("-" * 30)
for term, freq in sorted_terms[:10]:
    print(f"{term:<15} | {freq:<10}")
```

##4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

1. **Read documents:** Load input text files.
2. **Lowercase:** Convert all text to lowercase.
3. **Remove punctuation:** Clean out commas, periods, and special characters.
4. **Tokenize:** Split text into individual words.
5. **Calculate TF:** Count word frequency divided by total words in the document.
6. **Calculate DF:** Count how many documents contain each word.
7. **Calculate IDF:** Use the formula `log10(Total Docs / DF)`.
8. **Calculate TF-IDF:** Multiply `TF * IDF` for every word in every document.
9. **Display results:** Print Term, TF, DF, IDF, and TF-IDF at each stage.
10. **Top 10 Terms:** Find the top 10 highest TF-IDF scoring terms per document.

```python
import math
import string


# STEP 1: Read multiple text documents

docs = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

cleaned_docs = {}
for name, text in docs.items():
    # STEP 2: Convert text to lowercase
    lower_text = text.lower()

    # STEP 3: Remove punctuation
    clean_text = lower_text.translate(str.maketrans('', '', string.punctuation))

    # STEP 4: Tokenize the documents
    tokens = clean_text.split()
    cleaned_docs[name] = tokens

print("--- Step 1 to 4: Preprocessed Tokens ---")
for name, tokens in cleaned_docs.items():
    print(f"{name}: {tokens}")


# STEP 5: Calculate TF for every term

tf_docs = {}
for name, tokens in cleaned_docs.items():
    doc_len = len(tokens)
    tf_docs[name] = {w: tokens.count(w) / doc_len for w in set(tokens)}


# STEP 6: Calculate DF for every term

all_words = set(w for tokens in cleaned_docs.values() for w in tokens)
N = len(docs)
df = {w: sum(1 for tokens in cleaned_docs.values() if w in tokens) for w in all_words}

# STEP 7: Calculate IDF using the logarithmic formula

idf = {w: math.log10(N / df[w]) for w in all_words}

# STEP 8 & STEP 9: Calculate TF-IDF & Display Term, TF, DF, IDF, TF-IDF

tfidf_results = {}
print("\n--- Step 8 & 9: Full Breakdown (Term, TF, DF, IDF, TF-IDF) ---")
for name, tfs in tf_docs.items():
    print(f"\n[{name}]")
    print(f"{'Term':<15} | {'TF':<6} | {'DF':<4} | {'IDF':<6} | {'TF-IDF':<6}")
    print("-" * 48)

    tfidf_results[name] = {}
    for w, tf_val in tfs.items():
        score = tf_val * idf[w]
        tfidf_results[name][w] = score
        print(f"{w:<15} | {tf_val:.3f}  | {df[w]:<4} | {idf[w]:.3f}  | {score:.3f}")

# STEP 10: Identify top 10 terms with highest TF-IDF score for each document

print("\n--- Step 10: Top Terms by Highest TF-IDF Score ---")
for name, scores in tfidf_results.items():
    top_terms = sorted(scores.items(), key=lambda x: x[1], reverse=True)[:10]
    print(f"\n{name}:")
    for term, score in top_terms:
        print(f"  {term:<15}: {score:.3f}")
```
