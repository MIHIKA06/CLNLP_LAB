## 5.1 Implement subword-level tokenization using a BPE-based tokenizer. Given input data (Drive link).
1. Tokenize the sentence into subword units.
2. Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)

**5.1 (a) Using pretrained model (GPT-2 BPE)**

```
from transformers import GPT2Tokenizer

tok = GPT2Tokenizer.from_pretrained('gpt2')
tokens = tok.tokenize(text)
ids = tok.convert_tokens_to_ids(tokens)

print("Total tokens:", len(tokens))
for t, i in list(zip(tokens, ids))[:50]:   # first 50 only
    print(f"{t:20} {i}")

# Note: "Ġ" = space before the word, "Ċ" = newline
```

**5.1 (b) Without using pretrained model (train BPE from scratch)**

```
from tokenizers import Tokenizer, models, trainers, pre_tokenizers

bpe = Tokenizer(models.BPE(unk_token="[UNK]"))
bpe.pre_tokenizer = pre_tokenizers.Whitespace()
trainer = trainers.BpeTrainer(vocab_size=1000, special_tokens=["[UNK]"])
bpe.train([DATA_PATH], trainer)

out = bpe.encode(text)
tokens, ids = out.tokens, out.ids

print("Total tokens:", len(tokens))
for t, i in list(zip(tokens, ids))[:50]:   # first 50 only
    print(f"{t:20} {i}")
```

## 5.2 Implement subword-level tokenization using a SentencePiece tokenizer. Given same input data of 5.1.
1. Tokenize the sentence into subword units.
2. Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)

**5.2 (a) Using pretrained model (T5 – SentencePiece)**

```
from transformers import T5Tokenizer

t5 = T5Tokenizer.from_pretrained('t5-small')
tokens = t5.tokenize(text)
ids = t5.convert_tokens_to_ids(tokens)

print("Total tokens:", len(tokens))
for t, i in list(zip(tokens, ids))[:50]:   # first 50 only
    print(f"{t:20} {i}")
```

**5.2 (b) Without using pretrained model (train SentencePiece from scratch)**

```
import sentencepiece as spm

spm.SentencePieceTrainer.train(
    input=DATA_PATH, model_prefix='my_sp', vocab_size=1000,
    model_type='bpe', hard_vocab_limit=False)

sp = spm.SentencePieceProcessor(model_file='my_sp.model')
tokens = sp.encode(text, out_type=str)
ids = sp.encode(text, out_type=int)

print("Total tokens:", len(tokens))
for t, i in list(zip(tokens, ids))[:50]:   # first 50 only
    print(f"{t:20} {i}")
```

## 5.3 Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
"The young student is reading an interesting book in the library."

(Implement two separate codes: a. Using Spacy and b. NLTK)

**5.3 (a) Using spaCy**

```
import spacy

nlp = spacy.load('en_core_web_sm')
sentence = "The young student is reading an interesting book in the library."
doc = nlp(sentence)

seen = set()
print(f"{'Token':12} {'POS':8} Description")
for t in doc:
    if t.text.lower() not in seen:
        seen.add(t.text.lower())
        print(f"{t.text:12} {t.pos_:8} {spacy.explain(t.pos_)}")
```

**5.3 (b) Using NLTK**

```
import nltk
for p in ['punkt', 'punkt_tab', 'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng']:
    nltk.download(p, quiet=True)

desc = {'DT': 'Determiner', 'JJ': 'Adjective', 'NN': 'Noun, singular', 'NNS': 'Noun, plural',
        'VBZ': 'Verb, 3rd person singular present', 'VBG': 'Verb, gerund/present participle',
        'IN': 'Preposition/subordinating conjunction', 'VBP': 'Verb, non-3rd person present',
        'VBD': 'Verb, past tense', 'RB': 'Adverb', 'PRP': 'Personal pronoun', '.': 'Punctuation'}

sentence = "The young student is reading an interesting book in the library."
tagged = nltk.pos_tag(nltk.word_tokenize(sentence))

seen = set()
print(f"{'Token':12} {'POS':6} Description")
for w, tag in tagged:
    if w.lower() not in seen:
        seen.add(w.lower())
        print(f"{w:12} {tag:6} {desc.get(tag, tag)}")
```

## 5.4 Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.
Input data: Same input data of 5.1

```
import spacy, pandas as pd
from collections import Counter

nlp = spacy.load('en_core_web_sm')
nlp.max_length = 2_000_000
doc = nlp(text)

counts = Counter((t.text, t.pos_) for t in doc if not t.is_space)
rows = [(w, pos, spacy.explain(pos), c) for (w, pos), c in counts.items()]
df = pd.DataFrame(rows, columns=['Token', 'POS', 'Description', 'Frequency'])
df.sort_values('Frequency', ascending=False).reset_index(drop=True)
```

## 5.5 Write a Python program to train a Word2Vec CBOW model using the given text corpus. Display the vector representation of a selected word and find the three most similar words.
Input corpus – create a file named corpus.txt (10 sentences given in the question).

```
corpus_text = """Natural language processing is interesting
Natural language processing is useful
I love machine learning
Machine learning is useful
Deep learning is a part of machine learning
Artificial intelligence is changing the world
Natural language processing uses machine learning
Deep learning and artificial intelligence are related
I study natural language processing
Machine learning helps artificial intelligence"""

with open('corpus.txt', 'w') as f:
    f.write(corpus_text)

from gensim.models import Word2Vec

sentences = [line.lower().split() for line in open('corpus.txt')]

cbow = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=0, epochs=200, seed=1, workers=1)

print("Vector for 'machine':\n", cbow.wv['machine'])
print("\nTop 3 similar to 'machine':")
for w, s in cbow.wv.most_similar('machine', topn=3):
    print(f"{w:15} {s:.4f}")
```

## 5.7 Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.
Input corpus: Use corpus.txt from Question 5.5.
Calculate similarity for: 1. machine and learning  2. natural and language  3. deep and learning

```
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=1, epochs=200, seed=1, workers=1)

pairs = [('machine', 'learning'), ('natural', 'language'), ('deep', 'learning')]
for a, b in pairs:
    print(f"Cosine similarity ({a}, {b}) = {model.wv.similarity(a, b):.4f}")
```
